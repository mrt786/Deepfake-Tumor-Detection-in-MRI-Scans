```python
import medigan
import torch
from PIL import Image
import matplotlib.pyplot as plt

# Load pretrained brain MRI model
model_id = 'brain_mri_progans'  # Or search: medigan.search('brain*')
model = medigan.load(model_id)

# Generate batch (e.g., 64 images)
images = model.generate(n_images=64, truncation=0.7)  # Truncation for quality

# Visualize grid
grid = medigan.utils.make_grid(images)
plt.imshow(grid.permute(1, 2, 0).numpy())
plt.axis('off')
plt.savefig('brain_tumor_synthetic.png')
plt.show()

# Save individuals
medigan.utils.save_images(images, 'synthetic_brain_tumors/')
```
